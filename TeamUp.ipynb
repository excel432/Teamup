```python
roles = pd.DataFrame([
    {"role_id": "R001", "project_name": "Campus Marketplace", "role_title": "UI/UX Designer",
     "required_skills": ["Wireframing", "Prototyping"], "hours_per_week": 6,
     "duration_weeks": 6, "target_level": "Intermediate",
     "topics": ["Education", "E-commerce"], "goal": "Build a portfolio", "work_mode": "Remote"},
    {"role_id": "R002", "project_name": "Campus Marketplace", "role_title": "Frontend Developer",
     "required_skills": ["HTML/CSS", "React", "UI Integration"], "hours_per_week": 8,
     "duration_weeks": 6, "target_level": "Intermediate",
     "topics": ["Education", "E-commerce"], "goal": "Build a portfolio", "work_mode": "Remote"},
    {"role_id": "R003", "project_name": "Health Info App", "role_title": "Data Analyst",
     "required_skills": ["Tableau", "Python", "Data Visualization"], "hours_per_week": 5,
     "duration_weeks": 5, "target_level": "Beginner",
     "topics": ["Healthcare", "Public Health"], "goal": "Solve a real-world problem", "work_mode": "Hybrid"},
])
print(roles)
```

```python
role = roles.iloc[0]
generic = {"Teamwork", "Communication", "Problem Solving", "Research"}

needed = [s for s in role["required_skills"] if s not in generic]

def skill_share(person_skills):
    have = [s for s in needed if s in person_skills]
    return len(have) / len(needed)

df["skill_share"] = df["skills"].apply(skill_share)
print(df[["user_id", "skills", "skill_share"]].head())
```

```python
df["hours_ok"] = df["hours_per_week"] >= role["hours_per_week"]
print(df[["user_id", "hours_per_week", "hours_ok"]].head())
print(df["hours_ok"].sum())
```

```python
def duration_ok(pref):
    if pref == "Flexible":
        return True
    is_short_project = role["duration_weeks"] <= 6
    return is_short_project == (pref == "Short-term")

df["duration_ok"] = df["collaboration_duration"].apply(duration_ok)
print(df[["user_id", "collaboration_duration", "duration_ok"]].head())
print(df["duration_ok"].sum())
```

```python
df["skills_ok"] = df["skill_share"] >= 0.5

df["fails"] = (~df["skills_ok"]).astype(int) + (~df["hours_ok"]).astype(int) + (~df["duration_ok"]).astype(int)

passed = df[df["fails"] == 0]
print(len(passed))
print(passed[["user_id", "professional_role", "skill_share", "hours_per_week"]])
```

```python
df["availability_score"] = (
    (df["hours_per_week"] / role["hours_per_week"]).clip(upper=1) + df["duration_ok"].astype(int)
) / 2

df["goals_score"] = df["collaboration_goals"].apply(lambda g: 1 if role["goal"] in g else 0)

print(df[["user_id", "availability_score", "goals_score"]].head())
```

```python
def prefs_score(row):
    topic_share = len(set(role["topics"]) & set(row["project_interests"])) / len(role["topics"])
    mode_match = 1 if row["preferred_work_mode"] == role["work_mode"] else 0
    return (topic_share + mode_match) / 2

levels = {"Beginner": 0, "Intermediate": 1, "Advanced": 2}

def level_score(person_level):
    gap = abs(levels[person_level] - levels[role["target_level"]])
    return {0: 1.0, 1: 0.5}.get(gap, 0.0)

df["prefs_score"] = df.apply(prefs_score, axis=1)
df["level_score"] = df["experience_level"].apply(level_score)

print(df[["user_id", "prefs_score", "level_score"]].head())
```

```python
df["score"] = 100 * (
    0.35 * df["skill_share"]
    + 0.25 * df["availability_score"]
    + 0.20 * df["goals_score"]
    + 0.15 * df["prefs_score"]
    + 0.05 * df["level_score"]
)

def label(row):
    if row["fails"] == 0:
        return "Strong match" if row["score"] >= 75 else "Good match" if row["score"] >= 50 else "No match"
    if row["fails"] == 1 and row["skill_share"] > 0:
        return "Partial match"
    return "No match"

df["label"] = df.apply(label, axis=1)

result = df[df["label"] != "No match"].sort_values("score", ascending=False)
print(result[["user_id", "professional_role", "score", "label"]])
```

```python
today = pd.Timestamp("2026-10-25")

activity = pd.DataFrame([
    {"user_id": "TU001", "last_task_update": "2026-10-24", "last_comment": "2026-10-20", "last_checkin": "2026-10-22", "away_until": None},
    {"user_id": "TU002", "last_task_update": "2026-10-15", "last_comment": "2026-10-17", "last_checkin": "2026-10-16", "away_until": None},
    {"user_id": "TU003", "last_task_update": "2026-10-10", "last_comment": "2026-10-12", "last_checkin": "2026-10-11", "away_until": "2026-10-30"},
    {"user_id": "TU004", "last_task_update": "2026-10-21", "last_comment": None, "last_checkin": None, "away_until": None},
])

for col in ["last_task_update", "last_comment", "last_checkin", "away_until"]:
    activity[col] = pd.to_datetime(activity[col])

print(activity)
```

```python
INACTIVE_AFTER_DAYS = 5
REASSIGN_AFTER_DAYS = 7

activity["last_activity"] = activity[["last_task_update", "last_comment", "last_checkin"]].max(axis=1)
activity["days_quiet"] = (today - activity["last_activity"]).dt.days

def status(row):
    if pd.notna(row["away_until"]) and row["away_until"] >= today:
        return "Away"
    if pd.isna(row["last_activity"]) or row["days_quiet"] >= REASSIGN_AFTER_DAYS:
        return "Needs reassignment"
    if row["days_quiet"] >= INACTIVE_AFTER_DAYS:
        return "Inactive"
    return "Active"

activity["status"] = activity.apply(status, axis=1)
print(activity[["user_id", "days_quiet", "status"]])
```

```python
def check_activity(members, today, inactive_after=5, reassign_after=7):
    df = members.copy()
    df["last_activity"] = df[["last_task_update", "last_comment", "last_checkin"]].max(axis=1)
    df["days_quiet"] = (today - df["last_activity"]).dt.days

    def status(row):
        if pd.notna(row["away_until"]) and row["away_until"] >= today:
            return "Away"
        if pd.isna(row["last_activity"]) or row["days_quiet"] >= reassign_after:
            return "Needs reassignment"
        if row["days_quiet"] >= inactive_after:
            return "Inactive"
        return "Active"

    df["status"] = df.apply(status, axis=1)
    return df[["user_id", "days_quiet", "status"]]

print(check_activity(activity, today))
```
